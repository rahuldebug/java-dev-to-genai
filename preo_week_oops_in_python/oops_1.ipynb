```python
class Employee:
    increment = 1.04
    number_of_employee = 0
    def __init__(self , fname , lname , salary):
        self.fname = fname
        self.lname = lname
        self.salary = salary
        Employee.number_of_employee +=1
    def apply_raise(self):
        self.salary= self.salary*Employee.increment


em1= Employee('rahul', 'choudhary', 5000000)
em1.apply_raise()
print(em1.salary)
print (em1.__dict__)
print(Employee.__dict__)
print(Employee.number_of_employee)
```

```python
import datetime

class Employee:
    increment = 1.05
    def __init__(self, fname, lname, salary):
        self.fname = fname
        self.lname = lname
        self.salary = salary

    @classmethod
    def create_class(cls, details):
        fname, lname, salary = details.split("_")
        return cls(fname, lname, salary)

    @staticmethod
    def is_weekday(day):
        if day.weekday() == 5 or day.weekday() == 6:
            return False
        return True

em1 = Employee.create_class("rahul_choudhary_5000000")
mydate = datetime.date(2026, 6, 29)
print(Employee.is_weekday(mydate))

            
```
