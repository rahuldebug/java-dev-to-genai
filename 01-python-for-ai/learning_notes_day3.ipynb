2. Understanding Decorators with Examples
A decorator in Python is a function that takes another function as an argument, adds some functionality, and returns it. It’s similar to Java annotations but more flexible.

```python
import time
def my_timer(func):
    def wrapper(*args, **kwargs):
        start_time=time.time()
        result=func(*args, **kwargs)
        end_time=time.time()
        print(f"Execution time of {func.__name__}: {end_time-start_time} seconds")
        return result
    return wrapper
@my_timer
def compute_square(n):
    return [i*i for i in range(n)]
print(compute_square(1000000))
```
